# 전기 자동차 기본 사항

1. 전기 자동차 기본 사항

플러그인 전기 자동차는 어떻게 작동합니까?

예, 공공 충전소는 슈퍼마켓 주차장, 시내 주차장, 주유소 및 전국의 기타 많은 장소에 있습니다. 일부 공공 충전소는 무료이며, 다른 충전소는 수수료 또는 멤버십이 필요합니다.

2. 전기차의 정의와 주요 장단점

전기차는 내연기관 대신 전기 모터를 동력원으로 사용하는 차량입니다. 배터리에 저
장된 전기를 사용하여 모터를 구동하며, 충전소를 통해 배터리를 재충전할 수 있습니
다. 전기차는 환경 친화적이며 배기가스를 배출하지 않는 것이 특징입니다

- 장점: 환경 친화적, 연료비 절감, 저소음, 유지보수 비용 절감, 성능 향상
- 단점: 주행 거리 제한, 충전 인프라 부족, 초기 비용 높음, 배터리 수명 문제

3. 전기 자동차를 구매하기 전에 무엇을 고려해야합니까?

운전 습관과 개인적 선호도 등 다양한 요인이 있습니다. 
- 총 주행 거리:얼마나 많이 주행하세요?
- 가솔린 사용:배터리 용량은 가솔린 한 방울을 사용하지 않고도 얼마나 멀리 갈 수 있는지 결정합니다
- 충전: 어디서 충전하나요? 얼마나 가요?여러 배터리 전기 자동차가 100~200마일 이상         주행가능합니다.
- 보험 비용:차종,운전전자 이력에 따라 보험사를 권장합니다.

4. EV가 더 저렴합니까?

구매 가격은 제조업체, 모델, 연도, 차량 유형 및 시장 부문(대규모 시장, 럭셔리 등)에 따라 가스 구동 차량과 같이 매우 다양합니다.배터리 크기에 의해 결정되므로 배터리가격에 따라 변동됩니다.

5. 전기 자동차의 일반적인 유지보수 비용은 얼마입니까?

플러그인 전기 자동차는 일반적으로 총 소유 비용이 낮고, 특히 유지보수 비용이 낮습니다. 이는 움직이는 부품이 적고, 오일 교환이 감소하고(또는 완전 전기가 아닌 경우), 브레이크 작업이 적기 때문입니다. 배터리 재생은 대부분의 에너지를 흡수합니다. 하이브리드 및 플러그인 전기 자동차는 브레이크 작업을 받기 전에 100,000마일을 주행가능합니다.

6. 총 소유권 절감액: 전기차 수명 기간 동안 얼마나 절약할 수 있습니까?

 전기차는 구매 비용이 10% ~ 40% 더 높을 수 있지만, 운전자가 구매 가격, 연료 공급 비용 및 유지보수 비용을 부담하면 전기차의 수명 기간 동안 총소유비용 절감이 됩니다.
 지방자치단체별로 다르지만 1000만원 가량 보조금도 지원되며 세금 감면혜택을 받을수 있습니다. 감가상각이 예상보다 적습니다.

8. 기대 수명을 초과한 배터리는 어떻게 됩니까?

- 배터리 수명:  제조사의 보증 기간이 보통 8년 또는 160,000km입니다.                          실제로는 더 오래 사용할 수 있습니다
- 유지보수: : 내연기관차 대비 부품이 적고 간단하여 유지보수 비용이 저렴합니다.

9. 전기차와 하이브리드차의 차이

EV의 배터리는 가스 구동 차량의 엔진과 유사합니다. EV의 예상 수명은 200,000마일로 추정됩니다. 배터리 수명이 초과되면 재활용될 수 있습니다
- 전기차: 배터리만을 동력원으로 사용하는 차량
- 하이브리드차: 전기 모터와 내연기관을 함께 사용하는 차량
플러그인 하이브리드차: 외부 전원으로 배터리를 충전할 수 있는 하이브리드차
