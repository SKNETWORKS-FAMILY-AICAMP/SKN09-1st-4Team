2. 전기 자동차의 환경적 이점은 무엇입니까?

전기 자동차는 우리가 연소하는 가솔린의 양을 줄이고 유지 보수 비용이 저렴합니다.

①작동 배기가스 감소:
플러그인 전기 자동차의 전기 구동 트레인과 관련된 배기가스는 테일 파이프 배기가스가 아닌 배터리를 충전하기 위해 전기를 생산하는 발전소에서 발생합니다. 또한 전기 자동차는 유정에서 휠에 이르기까지 내연기관 자동차에 비해 이산화탄소(CO2)를 상당히 적게 방출합니다. CO2는 지구 온난화와 관련된 주요 가스입니다.
수질 보존:석유 가솔린 및 엔진 오일의 사용이 감소하면 바다, 강 및 지하수에 대한 유출 및 오염이 줄어듭니다.

②소음 감소:
더 깨끗해지는 것 외에도 전기 자동차는 가솔린 동력 차량보다 조용해져 소음 오염이 줄어듭니다.

3. 전기 자동차의 보조금은 얼마나 되나요?

2025년 전기차 보조금 차량의 크기와 성능에 따라 차등 지급됩니다.
- 중형 전기 승용차: 최대 300만원
- 소형 전기 승용차: 최대 250만원
- 초소소형 전기 승용차: 최대 200만원

2024년 대비 중형차 100만원, 소형차 50만원 줄어든 금액입니다.

클릭시 전국 보조금 사이트 https://www.tior.co.kr/ev/subsidy/step1?car_type=11

4. 소비자 및 제조사 인센티브

- 청년층 추가 지원
   19세~34새 생애 첫 전기차 구매시 기본 보조금 20%지원됩니다.
   저소득층 쳥년 30% 추가 지원에서서 확대한 조치입니다.
- 제조사 가격 할인 인센티브
   5300만원 이하 차량 제조사 할인가격 제공, 추가 보조금이 지원됩니다.
   할인금액 500만원 이하일 경우 20% 추가 보조금
   할인금액이 500만원 초과일 경우 40% 추가 보조금금
- 상용차 보조금 강화
   전기 밴: 주행 거리 500km 미만 경우 보조금 차감 됩니다. 
   전기 스쿨 버스 : 최대 1억 1,500만원의 보조금 지원.
   전기 틀럭: 주행거리 280km 이상 및 고속 충전지원시 추가 보조금 혜택
   농업종사자 10% 추가 지원됩니다.
