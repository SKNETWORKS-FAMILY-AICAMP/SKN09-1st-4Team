# 요금 및 비용

1. 가정, 직장 또는 공공 충전소에 플러그를 꽂으면 어떻게 전기 요금이 청구됩니까?

   가정에서 전기차를 충전할 때 사용되는 전기는 월간 공과금 청구서에 요금으로 표시됩니다. 직장에 충전소가 있는 경우, 충전기의 전기가 어떻게 청구되는지 고용주에게 문의하십시오. 공공 충전소의 경우, 전기 가격은 각 충전 네트워크 제공업체에 따라 다릅니다.

계정을 만들고 선호하는 결제 방법을 추가하면 네트워크 공급자의 충전소를 사용하여 전기차를 충전하고 전기차의 플러그를 뽑은 후 사용한 전기 요금을 지불할 수 있습니다. 충전 네트워크 제공업체는 운전자에게 kWh 요율로 월 구독료를 지불하거나 각 충전 세션 후 제공업체의 현재 kWh 요율을 지불할 수 있는 옵션을 제공합니다.

2. 공용 충전소는 어떻게 선택합니까?

Tesla 소유자의 경우, 공용 Tesla 충전소를 사용할 수 있습니다. Tesla의 충전소 지도를 사용하여 근처의 충전기를 찾아보십시오. 또한 Tesla 드라이버는 Tesla 플러그 어댑터가 있는 다른 충전 네트워크 제공업체의 충전기를 사용할 수 있습니다. 네트워크 공급자에 계정을 생성해야 합니다.

다른 EV 소유자의 경우 – 테슬라 이외의 공공 충전소를 사용할 수 있는 옵션이 있습니다. 공용 충전기를 사용하려면 온라인 또는 모바일 앱을 다운로드하여 충전 네트워크 제공업체에 계정을 만들어야 합니다. 귀하의 운전 요구에 가장 적합한 네트워크 서비스 제공자를 선택하십시오. 결정을 내릴 때, 일일 통근 경로, 주말에 가는 장소, 장거리 도로 여행을 고려해야 합니다.

3. 공용 충전소는 어떻게 선택합니까?

전기차 채택 및 충전이 증가하면 실제로 전기 비용을 절감하는 데 도움이 될 수 있습니다.         이는 전기 그리드 업그레이드 및 유지보수와 관련된 고정 비용이 추가된 전기 사용량에 분산될 수 있기 때문에 그리드를 사용하는 모든 고객의 전기 비용을 낮추는 데 도움이 됩니다. 이는 EV 운전자가 사용 시간(TOU) 요율에 등록하고 비수기 시간에 차량을 충전할 때 특히 그렇습니다.
