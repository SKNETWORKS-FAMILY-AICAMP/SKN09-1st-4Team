2. 전기 자동차의 이점은 무엇입니까?

전기 자동차는 우리가 연소하는 가솔린의 양을 줄이고 유지 보수 비용이 저렴합니다.
-작동 배기가스 감소:
플러그인 전기 자동차의 전기 구동 트레인과 관련된 배기가스는 테일 파이프 배기가스가 아닌 배터리를 충전하기 위해 전기를 생산하는 발전소에서 발생합니다. 또한 전기 자동차는 유정에서 휠에 이르기까지 내연기관 자동차에 비해 이산화탄소(CO2)를 상당히 적게 방출합니다. CO2는 지구 온난화와 관련된 주요 가스입니다.
수질 보존:석유 가솔린 및 엔진 오일의 사용이 감소하면 바다, 강 및 지하수에 대한 유출 및 오염이 줄어듭니다.

- 소음 감소:
더 깨끗해지는 것 외에도 전기 자동차는 가솔린 동력 차량보다 조용해져 소음 오염이 줄어듭니다.

- 정부 보조금: 
국가와 지자체에서 전기차 구매 보조금을 지원합니다.
세금 감면: 취득세, 등록세 등 세금이 감면될 수 있습니다

- 기타 혜택: 
충전소 설치 지원, 주차장 요금 할인 등의 혜택이 있습니다.

- 충전 인프라: 
지속적으로 확장되고 있지만, 일부 지역은 충전소 접근성이 낮습니다.

6. EV 배터리 상태를 연장하려면 어떻게 해야 합니까?

EV의 배터리 상태를 연장하는 방법에 대한 팁과 함께 몇 가지 예입니다.

- 과충전:
   과충전하거나 전기차 충전량을 완전히 고갈시키지 마십시오. EV는 극한 상태에서 충전되거나 방전되는 것을 방지하는 데 도움이 되는 배터리 관리 시스템과 함께 설치됩니다. 이상적으로는 20~80%의 충전량을 유지하고 장거리 주행 시에만 완전히 충전하는 것이 좋습니다.
- 온도:
   배터리에는 온도 제어 시스템이 내장되어 있지만, 주차하는 동안 극심한 열에 노출되는 것을 최소화해야 합니다. 더운 날에 주차할 수 있는 그늘진 장소나 차고를 찾으십시오
- 고속 충전:
   빠른 충전기 사용을 제한하십시오. 전기차를 충전하기 위해 고속 충전기에만 의존하지 않는 것이 가장 좋습니다. 전기차가 짧은 시간 내에 전기차 배터리에 많은 전류를 가하기 때문입니다. 이렇게 하면 시간이 지남에 따라 배터리가 스트레스를 받고 품질이 더 빠르게 저하됩니다. 충전 속도가 느려질 때마다 레벨 1 또는 레벨 2 충전기를 사용하십시오.

7. 배터리가 고장나면 어떻게 되나요?

EV 배터리 문제는 다행히 드물다. EV 배터리는 다양한 모듈로 설계되어 문제가 발생할 경우 더 안전하고 쉽게 교체할 수 있습니다. 보증에 따라 고장이 발생하고 보증이 무효화되지 않은 경우, 오토메이커는 배터리를 수리하거나 교체할 책임이 있습니다. 
차량이 10년 이상 되었거나 사고가 발생한 경우, 소유자(또는 보험)는 배터리 교체 비용을 지불해야 합니다. 배터리 비용은 자동차 제조업체와 배터리 크기에 따라 다르지만 비용이 많이 드는 수리가 될 수 있습니다.
